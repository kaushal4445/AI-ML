```python
campaigns = np.array([
[101, 10000, 800, 80, 5000, 7500],
[102, 15000, 1200, 150, 7000, 12000],
[103, 8000, 400, 30, 3000, 4200],
[104, 20000, 1800, 250, 9000, 18000],
[105, 12000, 600, 90, 6000, 8000],
[106, 18000, 1500, 220, 8000, 16000],
[107, 9000, 300, 20, 2500, 3000],
[108, 25000, 2200, 300, 11000, 24000]
])
```

```python
# Campaign_ID
# Impressions
# Clicks
# Conversions
# Cost
# Revenue
# CTR = Clicks / Impressions × 100
# Conversion Rate = Conversions / Clicks × 100
# ROI = (Revenue - Cost) / Cost × 100
# Find highest-revenue campaign.
# Find highest-ROI campaign.
# Calculate average CTR.
# Calculate average ROI.
# Find campaigns with above-average CTR and ROI.
```

```python
revenue = campaigns[:,-1]
revenue
```

```python
ind = np.argmax(revenue)
ind
```

```python
cost = campaigns[:,-2]
cost
```

```python
# Find highest-ROI campaign.
roi = ((revenue - cost)*100)/cost
roi
```

```python
high_roi = np.argmax(roi)
campaigns[high_roi]
```

```python
# Calculate average CTR.
impressions = campaigns[:,1]
clicks = campaigns[:,2]
ctr = (clicks*100)/impressions
ctr
```

```python
avg_ctr = np.mean(ctr)
avg_ctr
```

```python
# Calculate average ROI.
avg_roi = np.mean(roi)
avg_roi
```

```python
# Find campaigns with above-average CTR and ROI.
criteria = ((ctr > avg_ctr) & (roi > avg_roi))
criteria
```

```python
campaigns[criteria]
```

```python
# Categorize campaigns:
# ROI ≥ 80 → Excellent
# ROI ≥ 40 → Good
# Otherwise → Poor
# Calculate percentage of Excellent campaigns.
# Find campaigns above 75th percentile of revenue.
```

```python
category = np.where(roi >= 80, "Excellent", np.where(roi >= 40, "Good", "Poor"))
category
```

```python
perc  = (len(category[category == "Excellent"])*100)/len(category)
perc
```
