```python
# Problem statement: An array contains distinct numbers from 1 to n,
# but one number is missing. Find the missing number.
# Input: arr = [1, 2, 4, 5, 6], n = 6
# output: 3
```

```python
def find_missing(arr, n):
  expected_sum = n * (n + 1) // 2
  actual_sum = sum(arr)
  return expected_sum - actual_sum
arr = [1, 2, 4, 5, 6]
print(find_missing(arr, 6))
```

```python
def find_missing(arr, n):

  for i in range(1, n + 1):
    if i not in arr:
      return i

arr = [1, 2, 4, 5, 6]
print(find_missing(arr, 6))
```

```python
# Find Two Numbers Whose Sum Equals a Target
# Problem statement: Given an array and a target value,
# return the indices of two distinct elements whose sum equals the target.
# Assume exactly one valid pair exists.
# Input: arr = [2, 7, 11, 15], target = 9
# Output: [0,1]
```

```python
def two_sum(arr, target):
  seen = {}
  for i, num in enumerate(arr):
    required = target - num

    if required in seen:
      return [seen[required], i]

    seen[num] = i

  return []

arr = [2, 7, 11, 15]
print(two_sum(arr, 9))
```

```python
def is_palindrome(text):
  cleaned = ""
  for char in text.lower():
    if char.isalnum():
      cleaned += char
  print(cleaned)
  rev_s = cleaned[::-1]
  if cleaned == rev_s:
    print("Yes")
  else:
    print("No")
  print("*"*15)

is_palindrome("A man, a plan, a canal: Panama")
is_palindrome("Hello")
```

```python
# Problem statement: Two strings are anagrams
# if they contain the same characters with the same frequencies, regardless of order.
# Ignore spaces and case.
```

```python
# Input:
# s1: "Listen"
# s2: "Silent"
# output: "Yes"
```

```python
def are_anagrams(s1, s2):
  s1 = s1.replace(" ", "").lower()
  s2 = s2.replace(" ", "").lower()
  if len(s1) != len(s2):
    return False
  frequency = {}
  for char in s1:
    frequency[char] = frequency.get(char, 0) + 1
  for char in s2:
    if char not in frequency:
      return False
    frequency[char] -= 1
    if frequency[char] < 0:
      return False
  return True

print(are_anagrams("Listen", "Silent"))
print(are_anagrams("Hello", "World"))
```

```python
# Problem statement: Given a string, find the first character that appears exactly once.
# Return -1 if every character repeats.
# Input: "aabbcddee"
# Output: 'c'
# Input: "aabcddee"
# Output: 'b'
```

```python
def first_unique_character(text):
  frequency = {}
  for char in text:
    frequency[char] = frequency.get(char, 0) + 1
  for char in text:
    if frequency[char] == 1:
      return char
  return -1

print(first_unique_character("aabcddee"))
print(first_unique_character("aabb"))
```

```python
# Problem statement: Given a sentence, return its longest word.
# If multiple words have the same maximum length, return the first one.
# Input: "Python makes problem solving easier"
# Output: "problem"
```

```python
def longest_word(sentence):
    words = sentence.split()
    if not words:
        return ""
    longest = words[0]
    for word in words[1:]:
        if len(word) > len(longest):
            longest = word
    return longest
sentence = "Python makes problem solving easier"
print(longest_word(sentence))
```

```python
# Problem statement:
# Given two arrays sorted in ascending order,
# merge them into one sorted array without sorting the combined result afterward.
# Input: [1, 4, 7, 9] and [2, 3, 8, 10]
# Output: [1, 2, 3, 4, 7, 8, 9, 10]
```

```python
def merge_sorted_arrays(a, b):
    i = 0
    j = 0
    result = []
    while i < len(a) and j < len(b):
        if a[i] <= b[j]:
            result.append(a[i])
            i += 1
        else:
            result.append(b[j])
            j += 1
    while i < len(a):
        result.append(a[i])
        i += 1
    while j < len(b):
        result.append(b[j])
        j += 1
    return result
a = [1, 4, 7, 9]
b = [2, 3, 8, 10]
print(merge_sorted_arrays(a, b))
```
