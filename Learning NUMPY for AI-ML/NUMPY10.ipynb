```python
sale1  = np.array([
    120000, 135000, 1280000, 150000,
    160000, 172000, 165000, 180000,
    195000, 210000, 205000, 220000
])
```

```python
np.mean(sale1)
```

```python
np.median(sale1)
```

```python
np.argmax(sale1)
```

```python
# Create a forecast for the next 6 months using np.license() between:
# the minimum observed sales
# and 20% above the maximum observed sales.T

# Round the forecast values to the nearest integer.

# Combines the original sales and forecast into one 18- element array.
# Reshape the combined array into 3*6.
# calculate the average sales for each row.
# Identify values that are greater than the overall median.
```

```python
np.min(sale1)
```

```python
np.max(sale1)
```

```python
forecast = np.linspace(np.min(sale1), np.max(sale1), 6)
```

```python
r_s = np.round(forecast, 2)
```

```python
r_s
```

```python
sale1
```

```python
n_s = np.append(sale1, r_s)
```

```python
n_s
```

```python
ans = np.reshape(n_s, (3, 6))
```

```python
ans
```

```python
np.mean(ans, axis = 1)
```

```python
np.where(ans > np.median(ans), "Yes", "No")
```
