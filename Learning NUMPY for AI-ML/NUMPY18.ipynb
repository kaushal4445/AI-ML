# **Higher Order Function (HOF)**

```python
# HOF are functions that:
# 1- accepts another function as an argument
# 2- returns another function
```

```python
# add (a,b)
# mul (a,b)
# operation() --> hof
```

```python
def add(a,b):
  return a+b
def mul(a,b):
  return a*b
def operation(f,a,b):
  return f(a,b)
```

```python
ans = operation(add,3,5)
ans
```

```python
ans = operation(mul,3,5)
ans
```

```python
# Given a list of integers, find the largest element without using Python's built-in max() function.
# Input:  [12, 45, 7, 89, 34]
# Output: 89
```

```python
numbers = [12, 45, 7, 89, 34]
largest = numbers[0]
for num in numbers:
  if num > largest:
    largest = num
print("Largest element:", largest)
```

```python
# Given a list of distinct integers, find the second largest element without sorting the list.
# Input:  [10, 45, 20, 78, 34]
# Output: 45
```

```python
numbers = [10, 45, 20, 78, 34]
largest = float('-inf')
second_largest = float('-inf')
for num in numbers:
  if num > largest:
    second_largest = largest
    largest = num
  elif num > second_largest:
    second_largest = num
print(largest)
print(second_largest)
```

```python
# Given a list of integers, create a new list containing only unique elements.
# Maintain the original order of elements.
# Input:  [10, 20, 10, 30, 20, 40]
# output: [10,20,30,40]
```

```python
numbers = [10, 20, 10, 30, 20, 40]
unique = []
for num in numbers:
  if num not in unique:
    unique.append(num)
print("Unique elements:", unique)
```

```python
# Move All Zeros to the End
# Given a list of integers, move all zeros to the end of the list
#  while maintaining the relative order of the non-zero elements.
# Input:  [0, 5, 0, 3, 8, 0, 2]
# output: [5,3,8,2,0,0,0]
```

```python
numbers = [0, 5, 0, 3, 8, 0, 2]
result = []
zero_count = 0
for num in numbers:
  if num == 0:
    zero_count += 1
  else:
    result.append(num)
for i in range(zero_count):
      result.append(0)
print("Result:", result)
```

```python
# Given a list of numbers, calculate the average and print all elements that are greater than the average.
# (without numpy)
# Input:  [10, 20, 30, 40, 50]
# Output: [40, 50]
```

```python
numbers = [10, 20, 30, 40, 50]
total = 0
for num in numbers:
  total += num
  average = total / len(numbers)
  result = []
for num in numbers:
  if num > average:
    result.append(num)

print("Average:", average)
print("Elements greater than average:", result)
```

```python
# Given a list containing positive, negative and zero values, create two lists:
# Positive numbers
# Negative numbers
# Ignore zero.
# Input: [10, -5, 0, 8, -3, 12, -7]
# Positive: [10, 8, 12]
# Negative: [-5, -3, -7]
numbers = [10, -5, 0, 8, -3, 12, -7]
positive = []
negative = []
for num in numbers:
  if num > 0:
    positive.append(num)
  elif num < 0:
    negative.append(num)

print("Positive:", positive)
print("Negative:", negative)
```

```python
# Given two lists, find the elements that appear in both lists.
# List 1: [10, 20, 30, 40]
# List 2: [30, 40, 50, 60]
# ans: [30,40]
```

```python
list1 = [10, 20, 30, 40]
list2 = [30, 40, 50, 60]
common = []
for num in list1:
  if num in list2:
    common.append(num)
print("Common elements:", common)
```
