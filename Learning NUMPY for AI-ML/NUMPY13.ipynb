```python
# The coordiabtes of eight delivery locations are:

# Each row contains:

# [x, y]

# Calculatr distance from the origin using:

# distance = root(x^2 + y^2)
```

```python
import numpy as np
```

```python
points = np.array([
    [3,4],
    [5, 12],
    [8, 15],
    [7, 24],
    [9, 40],
    [12, 35],
    [15, 20],
    [6, 8]

])
```

```python
# Extract x and y coordinates.

# Claculate istance for every location.
# Round distance to 2 deciamls.
# Find nearest anf Farthest location.

# Find the average distance.

# Find locations above  average distance.

# Categorizes :
# if >=30 - long
# elseif >= 15 - median
# otherwise  - short

# Find the percentage of long -distance deliveries.

# Flatten the coordinate array.

# Find the size of the flattened array.
```

```python
# Extract x and y coordinates.
x = points[:, 0]
y = points[ :, 1]
x , y
```

```python
# Claculate istance for every location.
dist = np.sqrt(np.square(x) + np.square(y))
dist
#
```

```python
# Find nearest anf Farthest location.

np.min(dist)
```

```python
np.max(dist)
```

```python
np.mean(dist)
```

```python
# Find locations above  average distance.
points[dist > np.mean(dist)]
```

```python
# Categorizes :
# if >=30 - long
# elseif >= 15 - median
# otherwise  - short

category=np.where( dist >= 30, "Long", np.where( dist >= 15, "Median", "Short"))
```

```python
# Find the percentage of long -distance deliveries.
lond_dist = category[category == "Long"]
lond_dist
```

```python
m = len(lond_dist)
n = len(dist)
perc = (m*100)/n
perc
```
