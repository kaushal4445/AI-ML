```python
# A teacher has recordered marks of 10 students:

marks = np.array([72, 85, 91, 64, 78, 88, 95, 70, 82, 76])
```

```python
# find:
# Total marks
# Average marks
# Highest marks
# Lowest marks
# Difference between highest and lowest marks
# Index of the student with the highest marks
```

```python
np.sum(marks)
```

```python
np.mean(marks)
```

```python
np.max(marks)
```

```python
np.min(marks)
```

```python
mx = marks[0]
mxi =0
n = len(marks)
for i in range(n):
  if  mx < marks[i]:
    mx = marks[i]
    mxi = i

print(mx)
print(mxi)
```
