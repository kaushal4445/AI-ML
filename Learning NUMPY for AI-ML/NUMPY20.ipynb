```python
numbers1 = [1, 2, 3]
numbers2 = [10, 20]
```

```python
pair = []
for n1 in numbers1:
  for n2 in numbers2:
    pair.append((n1,n2))
pair
```

```python
#do with list comprehension
pair = [ (n1,n2) for n1 in numbers1 for n2 in numbers2 ]
pair
```

```python
# Generate all possible combinations of T-shirt colors and sizes
colors = ["Red", "Blue", "Black"]
sizes = ["S", "M", "L"]
products = [ [c,s] for c in colors for s in sizes]
products
```

```python
# Generate pairs whose sum is greater than 12
numbers1 = [1, 2, 3]
numbers2 = [10, 20]
ans = []
for n1 in numbers1:
  for n2 in numbers2:
    if n1+n2 > 12:
      ans.append((n1,n2))
ans
```

```python
ans = [ (n1,n2) for n1 in numbers1 for n2 in numbers2  if n1+n2 > 12 ]
ans
```

```python
# Find pairs of numbers from 1 to 5 where both numbers are different and their product is even.
ans = []
for n1 in range(1,6):
  for n2 in range(1,6):
    if (n1 != n2) and (n1*n2 %2 == 0):
      ans.append((n1,n2))
ans
```

```python
pairs = [ (n1,n2) for n1 in range(1,6) for n2 in range(1,6) if (n1 != n2) and (n1*n2 %2 == 0) ]
pairs
```

```python
marks = [
[75, 80, 90],
[60, 85, 70],
[95, 88, 92]
]
ans = []
for r in marks:
  for e in r:
    ans.append(e)
ans
```

```python
# Extract only passing marks from every row
# marks >= 75
marks = [
[75, 80, 90],
    [60, 85, 70],
    [95, 88, 92]
]
```

```python
flat = [ e for r in marks for e in r if e >= 75 ]
flat
```

```python
# Increase every number by 5
marks = [
    [75, 80, 90],
    [60, 85, 70],
    [95, 88, 92]
]
flat = [e+5 for r in marks for e in r ]
flat
```

```python
# Replace negative values with zero in a matrix
data = [
    [10, -5, 20],
    [-3, 15, -8],
    [25, 0, 12]
]
ans = [e if e>=0 else 0 for r in data for e in r ]
ans
```

```python
digits = [1, 2, 3]
# Generate all possible three-digit combinations
ans = [(a,b,c) for a in digits for b in digits for c in digits ]
print(ans)
```

```python
digits = [1, 2, 3]
# Generate all possible three-digit combinations
ans = [ (a,b,c)
        for a in digits
        for b in digits
        for c in digits
      ]

print(ans)
```

```python
# Generate combinations where all three values differ
digits = [1, 2, 3]
ans = [(a,b,c) for a in digits for b in digits for c in digits if a!=b and b!= c and a != c ]
ans
```
