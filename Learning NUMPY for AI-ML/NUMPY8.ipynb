```python
prices = np.array([
    499.50, 799.90, 1299.75,
    2499,50, 3999.90, 599.25,
    899.80, 1599.40
])
```

```python
np.round(prices)
```

```python
prices[prices > 1000]
```

```python
prices[prices > 1000] * 0.85
```

```python
np.mean(prices[prices > 1000] * 0.85)
```

```python
np.max(prices)
```

```python
np.argmax(prices)
```

```python
prices[-4:]
```

```python
sqrt_prices = np.sqrt(prices[prices > 1000] * 0.85)
final = np.round(sqrt_prices, 2)
print(final)
```
