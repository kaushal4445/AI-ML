```python
products = []
while True:
  print("1.add product")
  print("2. view all product with price")
  print("3. gst calcualte")
  print("4. discount >= 3000 10%, <=2000 6%")
  print("5. final bill")
  print("6. Exit")

  ch = int(input("Enter your choice = "))


  if ch == 1:
    product_name = input("Enter product name: ")
    quantity = int(input("Enter quantity: "))
    price = int(input("Enter price: "))
    products.append([product_name, quantity, price])
    print("Product added successfully!")
  elif ch == 2:
    if len(products) == 0:
      print("No product found")
    else:
      print("--- All Products ---")
      for product in products:
        print(f"Product: {product[0]}, Quantity: {product[1]}, Price: {product[2]}")
  elif ch == 3:
    if len(products) == 0:
        print("No products to calculate GST for.")
    else:
        print("--- GST Calculation ---")
        for item in products:
          gst = item[2] * (20/100)
          print(f"GST for {item[0]} is: {gst:.2f}")
  elif ch == 4:
    if len(products) == 0:
        print("No products to calculate discount for.")
    else:
        print("--- Discount Calculation ---")
        for item in products:
          discount = 0
          if item[2] >= 3000:
            discount = item[2] * (10/100)
          elif item[2] <= 2000:
            discount = item[2] * (6/100)
          print(f"Discount for {item[0]} (price: {item[2]}) is: {discount:.2f}")
  elif ch == 5:
    if len(products) == 0:
        print("No products in the bill.")
    else:
        print("--- Final Bill ---")
        total_bill = 0
        for item in products:
            item_total = item[1] * item[2]
            gst = item_total * (20/100)
            discount = 0
            if item[2] >= 3000:
                discount = item_total * (10/100)
            elif item[2] <= 2000:
                discount = item_total * (6/100)
            final_price = item_total + gst - discount
            total_bill += final_price
            print(f"Product: {item[0]}, Qty: {item[1]}, Price: {item[2]}, Item Total: {item_total:.2f}, GST: {gst:.2f}, Discount: {discount:.2f}, Final Item Price: {final_price:.2f}")
        print(f"\nTotal Bill Amount: {total_bill:.2f}")
  elif ch == 6:
    print("Exiting application. Goodbye!")
    break
  else:
    print("Invalid choice. Please try again.")
```

```python
products = []

while True:

    print("1. Add Product")
    print("2. View Product")
    print("3. Calculate GST")
    print("4. Calculate Discount")
    print("5. Final Bill")
    print("6. Exit")

    choice = int(input("Enter your choice = "))

    if choice == 1:

        name = input("Enter product name = ")
        quantity = int(input("Enter quantity = "))
        price = int(input("Enter price = "))

        products.append([name, quantity, price])

        print("Product added Successfully")

    elif choice == 2:

        if len(products) == 0:
            print("No Product Found")

        else:
            for product in products:
                print("Name", product[0],
                      "Quantity", product[1],
                      "Price", product[2])

    elif choice == 3:

        if len(products) == 0:
            print("No Product Found")

        else:
            for product in products:

                total = product[1] * product[2]
                gst = total * 20 / 100

                print("Name", product[0])
                print("GST", gst)

    elif choice == 4:

        if len(products) == 0:
            print("No Product Found")

        else:
            for product in products:

                total = product[1] * product[2]

                if total >= 3000:
                    discount = total * 10 / 100

                elif total <= 2000:
                    discount = total * 6 / 100

                else:
                    discount = 0

                print("Name", product[0])
                print("Discount", discount)

    elif choice == 5:

        if len(products) == 0:
            print("No Product Found")

        else:

            total_bill = 0

            for product in products:

                total = product[1] * product[2]

                gst = total * 20 / 100

                if total >= 3000:
                    discount = total * 10 / 100

                elif total <= 2000:
                    discount = total * 6 / 100

                else:
                    discount = 0

                final_price = total + gst - discount

                total_bill = total_bill + final_price

                print("Name", product[0])
                print("Quantity", product[1])
                print("Price", product[2])
                print("Total", total)
                print("GST", gst)
                print("Discount", discount)
                print("Final Price", final_price)

            print("Total Bill =", total_bill)

    elif choice == 6:

        print("Exit")
        break

    else:

        print("Invalid Choice")
```
