```python
class Rectangle:
  def __init__(self, length, width):
    self.length = length
    self.width = width

  def area(self):
    return self.length * self.width


r = Rectangle(5, 10)
print(r.area())
```

```python
class Circle:

  def __init__(self, radius):
    self.radius = radius

  def area(self):
    return 3.14 * self.radius * self.radius


c = Circle(5)
print(c.area())
```

```python
class Bank:

  def __init__(self, balance):
    self.balance = balance

  def deposit(self, amount):
    self.balance += amount

  def withdraw(self, amount):
    self.balance -= amount

account = Bank(1000)

account.deposit(500)

print(account.balance)
```

```python
class Student:

  def __init__(self, name, age, marks):
    self.name = name
    self.age = age
    self.marks= marks

  def get_grade(self):
    if self.marks >= 40:
      return "Pass"
    else:
      return "Fail"


s1 = Student("Raj", 22, 45)
print(s1.get_grade())
```

```python
class Counter:
  def __init__(self):
    self.count = 0 # Initialize count to 0

  def increase(self): # Define a method to increase the count
    self.count += 1

c = Counter()

for i in range(5):
  c.increase()

print(c.count)
```

# Inheritance

```python
# parent class --- > the class we  are inheriting frim ---> base class

# child class----> the class that inhertis from another classs --> deriver class
```

```python
class Person:
  def __init__(self, first_name, last_name):
    self.first_name = first_name
    self.last_name = last_name


  def show_name(self):
    print(self.first_name, self.last_name)

```

```python
p1 = Person("Kaushal", "Mehra")
p1.show_name()
```

```python
class Student(Person):
  pass
```

```python
s1 = Student("Raj", "Sharma")
```

```python
s1.show_name()
```

```python
class Vehicle:
  def __init__(self, brand_name):
    self.brand_name = brand_name

  def display(self):
    print("Brand of Car is :" , self.brand_name)
```

```python
v1 = Vehicle("Maruti")
v1.display()
```

```python
class Car(Vehicle):
  pass
```

```python
c1 = Car("Swift")
c1.display()
```
