problem 2: word frequency counter
Problem Statement

Take  a sentnce as input.

Display how many times each word appears.



```python
s = input()
words =s.split(" ")
print(words)
```

```python
freq ={}
for word in words:
  if word not in freq:
    freq[word] = 1
  else:
    freq[word] += 1

print(freq)
```

Problem 4: Character Counter

Take a string as input

Display

Number of vowels

Number of cononants

Number of digits

Number of spaces


```python
s = input()
```

```python
s1 =s.lower()
print(s1)
```

```python
vowels = 0
consonants =0
digits = 0
spaces =0

for c in s1:
  if c in 'aeiou':
    vowels += 1
  elif c.isalpha():
    consonants += 1
  elif c.isdigit():
    digits += 1
  elif c.isspace():
    spaces += 1

print("Vowels -", vowels)
print("Consonants - ", consonants)
print("Digits -",  digits)
print("Spaces", spaces)

```
