```python
numbers =[10, 15, 20]
result = list(map(lambda x : x /5 , numbers))
print(result)
```

```python
numbers =[10, 15, 20]
result = list(map(lambda x : x //5 , numbers))
print(result)
```

```python
numbers1 = [3, 6, 9, 12]
result1 = list(map(lambda x : x +1, filter(lambda x : x % 6 == 0, numbers1)))
print(result1)
```

```python
marks = [35, 45, 80, 25, 90]
passed = filter(lambda x : x >= 40, marks)
bonus = list(map(lambda x : x + 5, passed))
print(bonus)
```

```python
numb = [2, 3, 4]
res = list(map(lambda x : x**2, filter(lambda x : x > 2,  numb)))
print(res)
```

```python

nu = [1, 2, 3, 4, 5]

re = list(map(lambda x : x*10, filter(lambda x: x%2 != 0, nu)))
print(re)

```

# **CSV**

```python
import csv
with open("student.csv", "w", newline = "") as file:
  writer = csv.writer(file)
  writer.writerow(["Name", "Marks", "City"])
  writer.writerow(["Raj", 95, "Mumbai"])
  writer.writerow(["koko", 90, "Hyderbad"])
```

```python
with open("student.csv", "r") as file:
  reader = csv.reader(file)
  for row in reader:
    print(row)
```
