# **Problem 6: Password Validator**

password is valid if it contains

minimum 8 charcters

at least one uppercase letter

at least one lowercase letter

at least one digit

Display  

Strong Password

or

Weak Password

```python
password = input("Enter your password: ")
print(password)

upper = False
lower = False
digit = False

n = len(password)
for char in password:
    if char.isupper():
      upper = True
    elif char.islower():
      lower = True
    elif char.isdigit():
      digit = True

if n >= 8 and upper == True and lower == True and  digit == True:
  print("Strong Password")
else:
  print("Weak Password")
```

# **Problem 7 : Cricket Scoreboard**

Take player names and runs.

Store them in a dictionary.

Display

Highest scorer

Average scorer

```python
scores = {
    "Virat" : 90,
    "Rohit" : 95,
    "Dhoni" : 100,
    "Rishabh Pant" : 50,
    "Hardik Pandya" : 20
}

type(scores)
```

```python
mx_score = 0
total_score = 0
for key in scores:
  value =scores[key]
  if value > mx_score:
    mx_score = value
    mx_player = key

  total_score += value


print("Average score", total_score/len(scores))


print("Max score", mx_score)
print("Max player", mx_player)
```

```python
player = max(scores, key = scores.get)
```

## **Problem 8; Common Subjects**

Input two sets of subjects.

Display

Common subjects

Subjects only in First student

Subjects only in second student


```python
s1 = {"Maths", "Science", "English"}
s2 = {"History", "Maths", "Science"}

common_subjects  = s1.intersection(s2)
# s1 & s2
print(common_subjects)

only_in_s1 = s1.difference(s2)
# s1- s2
print(only_in_s1)

only_in_s2 = s2.difference(s1)
# s2 - s1
print(only_in_s2)
```

#Problem 9: Most Frequent Character

find the charcter that appears the maximum number of times ina string.

Ignore spaces.

```python
s = "Pyhton is the most amazing programming language for ML today"
frq ={}
for c in s:
  if c != " ": # Changed from != "" to != " " to ignore spaces
    if c not in frq:
      frq[c] = 1 # Corrected indentation
    else:
      frq[c] += 1

print(frq)
```
