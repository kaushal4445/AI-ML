# **Map**

```python
# map(Fuction, iterable) ---> it returns map unction
```

```python
# def cube(lst):
#   cb = []
#   for n in lst:
#     cb.append(n**3)

#   return cb
```

```python
# cube([2, 4, 6])
```

```python
def cb(n):
  return n**3

cube = []
lst = [4, 6, 2]
for elem in lst:
  cube.append(cb(elem))

print(cube)
```

```python
lst = [4, 6, 2]
cube1 = list(map(lambda x : x**3, lst))
print(cube1)
```

```python

cube1 = list(map(lambda x : x**3,[4, 6, 2] ))
print(cube1)
```

## **Filter**

```python
num = [1, 2, 3, 4, 5 ,6, 7]
even = []
for n in num:
  if n % 2 == 0:
    even.append(n)
print(even)
```

```python
num = [1, 2, 3, 4, 5 ,6, 7]
even_numbers = list(filter(lambda x:  x%2==0, num))
print(even_numbers)
```

```python
even_numbers = list(filter(lambda x:  x%2!=0, num))
print(even_numbers)
```
